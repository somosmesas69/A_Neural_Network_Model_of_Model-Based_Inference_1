# **Model Analysis Tutorial**

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

### **The Two Step Environment**

```python
'''
Initializing the Two Step Environment and checking that the block switches occur
'''

import Two_Step_Environment as env

records = []
two_step = env.Naive_Two_Step()

for trial in range(1000):                                           # 1000 trials
    choice = np.random.choice(["Right","Left"])                     # Choose randomly
    state, reward, high_state = two_step.a_trial(choice)

    records.append( { 
        "trial":trial,
        "choice":choice,
        "state":state,
        "reward": reward,
        "high_state": high_state
    })

my_df = pd.DataFrame(records)
my_df["high_state_in_num_form"] = (my_df["high_state"] == "A").astype(int)
fig, ax = plt.subplots(1,1)
ax.plot(my_df["trial"],my_df["high_state_in_num_form"])
ax.set(title= "Trials vs High_State", ylabel = "High_State, A = 1", xlabel = "Trial #")
fig.suptitle("Are Block Switches Occuring?")
```
